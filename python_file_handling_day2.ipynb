```python
def listcount(listval):
  my_dict={}
  for i in listval:
    if i in listval:
      i=i.lower()
      my_dict[i]=my_dict.get(i,0)+1
  return my_dict
     
```

```python
import math
try:
    with open('notest.txt', 'r') as file:
        list_lines=[]
        lines = file.readlines()
        for line in lines:
                list_lines.append(line.strip())
        print("Values from text file:",list_lines)
        
        int_list = [int(item) for item in list_lines]
        print("Converting string to integer in same list",int_list)
        
        total_sum=math.fsum(int_list)
        print("Total Sum:",total_sum)
        
        total_avg=total_sum/len(int_list)
        print("Total Avg:",total_avg)
        
        print("Maximum from List:",max(int_list))
except FileNotFoundError:
    print("Error: The file 'notest.txt' was not found.")
```

```python
#Read a file and reverse each line.
try:
   
        with open('uppercasenotes.txt','r') as r:
            con=r.read()
            text="".join(reversed(con))
            print(text)

except FileNotFoundError:
    print("Error: The file 'uppercasenotes.txt' was not found.")
```

```python
#Sort lines in a file alphabetically.
try:
   
        with open('uppercasenotes.txt', "r+") as f:
            lines = f.read().splitlines()
        lines.sort()
        with open('writing.txt', "w") as w:
            w.write("\n".join(lines))

except FileNotFoundError:
    print("Error: The file 'uppercasenotes.txt' was not found.")
```

```python
#Read names from a file and remove duplicate names.
try:
   
        with open('uppercasenotes.txt', "r+") as f:
            lines = f.read().splitlines()
        lines.sort()
        print(lines)
        set_wodup=set(lines)
        print(set_wodup)
      
except FileNotFoundError:
    print("Error: The file 'uppercasenotes.txt' was not found.")
```
