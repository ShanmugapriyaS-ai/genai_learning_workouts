```python
#Reverse the string function - method 2
def reverslicing(text):
  return text[::-1]
```

```python
#Reverse the string
print(reverslicing("Hello i am priya"))
```

```python
#Check the Palindrome
str_p1="malayalam"
str_p2=reverslicing(str_p1)
if(str_p1==str_p2):
  print(f"The given String {str_p1} is palindrome")
else:
  print(f"The given String {str_p2} is not palindrome")
```

```python
#Vowel function
def vowelcheck(text):
  my_dict={}
  vowel={'a','e','i','o','u'}
  for i in text:
    if i in vowel:
      my_dict[i]=my_dict.get(i,0)+1
  return my_dict
```

```python
#Count the vowel in string
print(vowelcheck("Shanmugapriya"))
```

```python
# Function for count each element from the list
def listcount(listval):
  my_dict={}
  for i in listval:
    if i in listval:
      my_dict[i]=my_dict.get(i,0)+1
  return my_dict
```

```python
#Count each element from the list
list_val=[1,1,2,3,3,3,3,3,5,5,5,5,5,5,5,7,7,7,7,7]
print("list element count:",listcount(list_val))
```
